```python
import random

print("Choose a level of difficulty:")
print("1. Easy (1-10)")
print("2. Medium (1-50)")
print("3. Hard (1-100)")
level = int(input("Enter your choice (1-3): "))

if level == 1:
    max = 10
elif level == 2:
    max = 50
elif level == 3:
    max = 100
else:
    print("invalid choice")
    exit()

random_number = random.randint(1, max)
i=1
while i<=max:
    guess = int(input("Guess the number: "))
    if guess == random_number:
        print("You guessed the number.")
        break
    elif guess < random_number:
        print("Too low please try again.")
    else:
        print("Too high please try again.")
    i+=1
print("The number was:", random_number)
print("You guessed the number in", i, "tries.")






```
